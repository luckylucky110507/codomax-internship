```python
import pandas as pd
df = pd.read_csv(r"C:\Users\princ\OneDrive\Desktop\student-scores.csv")
```

```python
print(df.isnull().sum())
```

```python
df.fillna(df.mean(numeric_only=True), inplace=True)
```

```python
print("Duplicate rows:", df.duplicated().sum())
df.drop_duplicates(inplace=True)
```

```python
print(df.describe())
print(df.info())
```

```python
df.to_csv("student_scores_cleaned.csv", index=False)
```
