```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\princ\student_scores_cleaned.csv")
```

```python
#scatter plot
plt.scatter(df['weekly_self_study_hours'], df['math_score'])
plt.xlabel('Weekly Study Hours')
plt.ylabel('Math Score')
plt.title('Study Hours vs Math Score')
plt.show()
```

```python
#Bar Chart
df['gender'].value_counts().plot(kind='bar')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.title('Gender Distribution')
plt.show()
```

```python
#Line Chart
df_sorted = df.sort_values('weekly_self_study_hours')
plt.plot(df_sorted['weekly_self_study_hours'], df_sorted['math_score'])
plt.xlabel('Weekly Study Hours')
plt.ylabel('Math Score')
plt.title('Study Hours vs Math Score Trend')
plt.show()
```
